```python
order_sales_test2 = order_sales_test2.dropna()
print(order_sales_test2["order_date"].isna().sum())
```

```python
# 파생컬럼 order_month 추가

order_sales_test2["order_month"] = order_sales_test2["order_date"].dt.to_period("M").astype(str)
print(order_sales_test2.head())
```

```python
# completed 주문만 대상으로 집계
completed = order_sales_test2[order_sales_test2["order_status"] == "completed"].copy()

# products 병합 (카테고리 추가)
completed = completed.merge(
    products[["product_id", "product_name", "category"]],
    on="product_id",
    how="left",
    validate="many_to_one",
)

by_category = completed.groupby("category")["line_total"].sum().sort_values(ascending=False)
by_product  = completed.groupby("product_name")["line_total"].sum().sort_values(ascending=False).head(10)
by_month    = completed.groupby("order_month")["line_total"].sum().sort_index()
by_customer = completed.groupby("customer_id")["line_total"].sum().sort_values(ascending=False).head(10)

print(by_category)
print(by_month)
print(by_product)
print(by_customer)
```

```python
# 총합 일치 검증: 원본 합계 == 각 축(groupby)별 합계의 합
base_total     = completed["line_total"].sum()
category_total = completed.groupby("category")["line_total"].sum().sum()
month_total    = completed.groupby("order_month")["line_total"].sum().sum()
customer_total = completed.groupby("customer_id")["line_total"].sum().sum()

print("원본 completed line_total 합계 :", base_total)
print("카테고리별 합계의 합          :", category_total)
print("월별 합계의 합                :", month_total)
print("고객별 합계의 합              :", customer_total)

assert base_total == category_total == month_total == customer_total
print("검증 통과: 모든 축의 합계가", base_total, "으로 일치")
```

```python
# 6. LLM pandas 코드 검증
# Prompt: "orders, order_items, products 로 completed 주문의 카테고리별 매출 합계 pandas 코드 줘"

# --- LLM 제안 원본 코드 (검증 대상) ---
llm = order_items.merge(products, on="product_id", how="left")
llm_result = llm.groupby("category")["price"].sum().sort_values(ascending=False)
print("[LLM 원본] products.price 합계, completed 필터 없음, validate 없음")
print(llm_result)
print("LLM 원본 합계:", llm_result.sum())

# --- 실제 데이터와 대조해 발견한 문제 ---
# 1) products.price(정가)를 매출로 사용 -> 실제 매출은 quantity * unit_price (line_total)
# 2) order_status == "completed" 범위 필터 없음 -> cancelled/refunded 까지 섞임
# 3) merge validate 미지정 -> 다대일 관계 미검증

# --- 검증 후 수정한 코드 ---
fixed = completed.groupby("category")["line_total"].sum().sort_values(ascending=False)
print("\n[수정본] completed + line_total(quantity*unit_price) + validate='many_to_one'")
print(fixed)
print("수정본 합계:", fixed.sum())

print("\n합계 차이 (LLM 원본 - 수정본):", llm_result.sum() - fixed.sum())
# 최종 판단: 수정 후 사용. 코드가 에러 없이 돌아도 컬럼/범위가 틀리면 답은 틀리다.
```
